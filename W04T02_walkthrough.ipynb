## 3. Missing keys, and `.get()`

`record["missing"]` raises `KeyError`. `.get()` does not — it gives you `None`, or a
default of your choosing.

````python
print(record.get("status"))          # OK
print(record.get("missing"))          # None
print(record.get("missing", "n/a"))    # n/a
````

## 5. `zip()` — building a dict from two lists

Sometimes your data starts as two separate lists — names in one, values in
another — instead of already paired up. `zip()` pairs them together, one from
each; `dict()` turns those pairs straight into a dictionary.

````python
hostnames = ["srv-01", "srv-02", "srv-03"]
disk_used = [87, 45, 120]

paired = list(zip(hostnames, disk_used))
print(paired)     # [('srv-01', 87), ('srv-02', 45), ('srv-03', 120)]

usage = dict(zip(hostnames, disk_used))
print(usage)                  # {'srv-01': 87, 'srv-02': 45, 'srv-03': 120}
print(usage["srv-02"])         # 45
````

```
hostnames:   srv-01     srv-02     srv-03
                │          │          │
disk_used:     87         45        120
                │          │          │
zip pairs:  (srv-01,87) (srv-02,45) (srv-03,120)
```

If the two lists are different lengths, `zip()` stops at the shorter one —
silently, with no error.

## 5. Sets — only unique values, in curly braces too

Created with `{a, b, c}` — or `set()` for an empty one. **Not** `{}`: that makes an
empty *dict*, not an empty set. No order, no indexing, and duplicates are dropped
automatically.

````python
labels = ["R-001", "R-002", "R-001", "R-003"]
unique_labels = set(labels)

print(unique_labels)          # duplicates gone
print(len(unique_labels))      # 3, not 4
print("R-001" in unique_labels)   # True - fast membership check

empty_set = set()    # NOT {} - that is an empty dict
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `record["missing"]` | `KeyError: 'missing'` | check with `.get()` or `in` first |
| `record[value]` when you meant `record["value"]` | `NameError` (or wrong result) | keys are strings — quote them |
| assuming key order matters for lookup | it doesn't — lookup is by name, not position | that is the entire point of a dict |
| `for value in record:` expecting key AND value | only gives you the keys | use `.items()` for both |
| `{}` expecting an empty set | it is an empty **dict** | use `set()` for an empty set |
| `zip()`-ing two lists of different lengths | silently stops at the shorter one, no error | check both lists are the same length first |
